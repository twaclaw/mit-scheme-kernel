# Plotting

`Scmutils` draws through X11, which a notebook cannot reach. The kernel records
what the plotting procedures draw and renders it with matplotlib as inline SVG.

Requirements: the `mechanics` executable on your `PATH`, and the kernel configured
to use it.

Use the plotting procedures exactly as in SICM, inside a `%%plot` cell.


## A function

`plot-function` samples between the given limits.

```scheme
%%plot --title "sin x" --xlabel "x" --grid
(define win (frame 0. 6.3 -1.2 1.2))
(plot-function win sin 0. 6.28 .02)
```

## A parametric curve

`plot-parametric` takes `t` to a pair `(x . y)`.

```scheme
%%plot --title "Lissajous 3:2"
(define win (frame -1.2 1.2 -1.2 1.2))
(plot-parametric win
  (lambda (t) (cons (sin (* 3 t)) (cos (* 2 t))))
  0. 6.3 .005)
```

## Data

`plot-xy` takes lists of abscissae and ordinates.

```scheme
%%plot --title "damped oscillation" --xlabel "t" --ylabel "x"
(define ts (iota 400 0. .025))
(define xs (map (lambda (t) (* (exp (- (* .3 t))) (cos (* 3 t)))) ts))
(define win (frame 0. 10. -1.1 1.1))
(plot-xy win ts xs)
```

## A trajectory from an integration

Plotting the phase portrait of a harmonic oscillator, integrated with
`state-advancer`.

```scheme
%%plot --title "phase portrait" --xlabel "q" --ylabel "p" --grid
(define ((L-harmonic m k) local)
  (- (* 1/2 m (square (velocity local)))
     (* 1/2 k (square (coordinate local)))))

(define (sysder m k) (Lagrangian->state-derivative (L-harmonic m k)))

(define win (frame -1.2 1.2 -1.2 1.2))
(let loop ((t 0.) (state (up 0. 1. 0.)))
  (if (< t 12.)
      (let ((next ((state-advancer sysder 1. 1.) state .05 1e-10)))
        (plot-point win (ref next 1) (ref next 2))
        (loop (+ t .05) next))))
```

## Notes

- Rendering is static: `graphics-clear` drops what came before, so an animation
  loop shows its final frame.
- A cell that creates two frames produces two figures.
- `%%plot` options: `--title`, `--xlabel`, `--ylabel`, `--width`, `--height`,
  `--grid`.
